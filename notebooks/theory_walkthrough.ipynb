# SVD Image Compression Theory

This executable walkthrough uses the `svdlab` library and real image data.

```python
import numpy as np

from svdlab.decompose import compute_svd, reconstruct, truncate_svd
from svdlab.io import load_image

image, info = load_image('sample', 'camera', 256)
gray = image[:, :, 0] if image.ndim == 3 else image
U, S, Vt = compute_svd(gray)
print(info)
print('shape=', gray.shape, 'rank=', len(S))
```

```python
Uk, Sk, Vtk = truncate_svd(U, S, Vt, 20)
reconstructed = reconstruct(Uk, Sk, Vtk)
print('maximum absolute error=', np.max(np.abs(gray - reconstructed)))
```

```python
from svdlab.metrics import psnr, ssim

print('PSNR=', psnr(gray, reconstructed))
print('SSIM=', ssim(gray, reconstructed))
```
